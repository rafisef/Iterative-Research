## 1. Input Validation
Pydantic schema enforces types, lengths, and formats before anything else touches the data.

```python
from pydantic import BaseModel, Field, field_validator
import re

class UserRegistration(BaseModel):
    username: str = Field(min_length=3, max_length=30)
    email: str
    age: int = Field(ge=13, le=120)

    @field_validator("username")
    @classmethod
    def username_alphanumeric(cls, v: str) -> str:
        if not re.fullmatch(r"[a-zA-Z0-9_]+", v):
            raise ValueError("Username must be alphanumeric")
        return v

    @field_validator("email")
    @classmethod
    def email_format(cls, v: str) -> str:
        if not re.fullmatch(r"[^@\s]+@[^@\s]+\.[^@\s]+", v):
            raise ValueError("Invalid email")
        return v

# --- Tests ---

user = UserRegistration(username="jake_42", email="jake@example.com", age=30)
assert user.username == "jake_42"
print("PASS: valid input accepted")

try:
    UserRegistration(username="drop;--", email="a@b.com", age=20)
    assert False, "Should have rejected malicious username"
except Exception:
    print("PASS: malicious username rejected")
```

## 2. Concurrency
`asyncio.Lock` prevents concurrent callers from stomping on shared state.

```python
import asyncio

class SafeCounter:
    def __init__(self):
        self._lock = asyncio.Lock()
        self._count = 0

    async def increment(self) -> int:
        async with self._lock:
            self._count += 1
            await asyncio.sleep(0)
            return self._count

    @property
    def value(self) -> int:
        return self._count

# --- Tests ---

N = 100
counter = SafeCounter()
await asyncio.gather(*(counter.increment() for _ in range(N)))
assert counter.value == N, f"Expected {N}, got {counter.value}"
print(f"PASS: counter reached {N} after {N} concurrent increments")
```

## 3. Information Leakage
Allowlist projection on user objects. Sensitive fields never leave the server.

```python
from dataclasses import dataclass

@dataclass
class InternalUser:
    id: str
    email: str
    password_hash: str
    mfa_secret: str
    role: str

def to_public_user(user: InternalUser) -> dict:
    return {"id": user.id, "email": user.email, "role": user.role}

# --- Tests ---

user = InternalUser("u-1", "a@b.com", "$argon2id$...", "JBSWY3DPEHPK3PXP", "admin")
public = to_public_user(user)

assert set(public.keys()) == {"id", "email", "role"}, f"Unexpected fields: {public.keys()}"
assert "password_hash" not in public and "mfa_secret" not in public
print("PASS: only allowlisted fields exposed, sensitive fields excluded")
```

## 4. Access Control
Numeric role hierarchy + resource ownership check. Non-admins can only touch their own resources.

```python
from typing import Literal

Role = Literal["admin", "manager", "user"]
ROLE_LEVEL: dict[Role, int] = {"admin": 3, "manager": 2, "user": 1}

def check_access(user_role: Role, user_id: str, resource_owner_id: str, required_role: Role) -> None:
    if ROLE_LEVEL[user_role] < ROLE_LEVEL[required_role]:
        raise PermissionError("Forbidden")
    if user_role != "admin" and resource_owner_id != user_id:
        raise PermissionError("Forbidden")

# --- Tests ---

check_access("admin", "u-1", "u-99", "admin")
print("PASS: admin accesses another user's resource")

try:
    check_access("user", "u-5", "u-99", "user")
    assert False, "Should have denied cross-user access"
except PermissionError:
    print("PASS: user denied access to another's resource")
```

## 5. Injection Vulnerabilities
Parameterized SQL queries. Command allowlist with `subprocess.run(shell=False)`.

```python
import sqlite3, subprocess, os

DB_PATH = "test_patterns.db"
if os.path.exists(DB_PATH): os.remove(DB_PATH)
conn = sqlite3.connect(DB_PATH)
conn.row_factory = sqlite3.Row
conn.execute("CREATE TABLE users (id TEXT PRIMARY KEY, email TEXT, role TEXT)")
conn.execute("INSERT INTO users VALUES ('u-1', 'alice@example.com', 'admin')")
conn.commit()

def find_user(conn: sqlite3.Connection, email: str) -> dict | None:
    row = conn.execute("SELECT id, email, role FROM users WHERE email = ?", (email,)).fetchone()
    return dict(row) if row else None

ALLOWED_COMMANDS = {"whoami": ["whoami"], "uptime": ["uptime"]}

def run_command(name: str) -> str:
    cmd = ALLOWED_COMMANDS.get(name)
    if cmd is None: raise ValueError(f"Command not allowed: {name}")
    return subprocess.run(cmd, capture_output=True, text=True, shell=False).stdout

# --- Tests ---

evil = find_user(conn, "' OR '1'='1")
assert evil is None, f"SQL injection returned data: {evil}"
print("PASS: SQL injection returned no results")

try:
    run_command("whoami; cat /etc/passwd")
    assert False, "Should have rejected injected command"
except ValueError:
    print("PASS: injected command name rejected")

conn.close()
```

## 6. Race Conditions (Optimistic Locking)
Update only succeeds if the row version hasn't changed. Stale reads get rejected.

```python
import sqlite3, os

DB_PATH = "test_patterns.db"
if os.path.exists(DB_PATH): os.remove(DB_PATH)
conn = sqlite3.connect(DB_PATH)
conn.execute("CREATE TABLE accounts (id TEXT PRIMARY KEY, balance REAL, version INTEGER DEFAULT 1)")
conn.execute("INSERT INTO accounts VALUES ('acct-A', 1000.0, 1)")
conn.execute("INSERT INTO accounts VALUES ('acct-B',  500.0, 1)")
conn.commit()

def transfer_funds(conn, from_id, to_id, amount, version):
    if amount <= 0: raise ValueError("Amount must be positive")
    cur = conn.execute(
        "UPDATE accounts SET balance = balance - ?, version = version + 1 WHERE id = ? AND version = ? AND balance >= ?",
        (amount, from_id, version, amount))
    if cur.rowcount == 0: raise RuntimeError("Transfer failed -- stale version or insufficient balance")
    conn.execute("UPDATE accounts SET balance = balance + ?, version = version + 1 WHERE id = ?", (amount, to_id))
    conn.commit()

# --- Tests ---

transfer_funds(conn, "acct-A", "acct-B", 200.0, version=1)
bal = conn.execute("SELECT balance FROM accounts WHERE id = 'acct-A'").fetchone()[0]
assert bal == 800.0, f"Expected 800, got {bal}"
print("PASS: valid transfer applied correctly")

try:
    transfer_funds(conn, "acct-A", "acct-B", 100.0, version=1)
    assert False, "Should have rejected stale version"
except RuntimeError:
    print("PASS: stale version rejected")

conn.close()
```

## 7. Error Handling
Expected errors (`AppError`) return clean messages. Unexpected errors get a generic 500.

```python
class AppError(Exception):
    def __init__(self, message: str, status_code: int = 500):
        super().__init__(message)
        self.status_code = status_code

def handle_error(err: Exception) -> dict:
    if isinstance(err, AppError):
        return {"status_code": err.status_code, "body": {"error": str(err)}}
    return {"status_code": 500, "body": {"error": "An unexpected error occurred"}}

# --- Tests ---

resp = handle_error(AppError("Not found", 404))
assert resp["status_code"] == 404 and resp["body"]["error"] == "Not found"
print("PASS: AppError returns correct status and message")

resp = handle_error(RuntimeError("Connection to secret-db.internal:5432 refused"))
assert resp["status_code"] == 500 and "secret-db" not in resp["body"]["error"]
print("PASS: unexpected error returns generic 500, no internals leaked")
```
